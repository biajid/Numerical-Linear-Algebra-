```python
def lincomb(c, v):
    n = len(v[0]);
    if n != len(c):
        raise ValueError('Size Mismatch');
    out_v = np.zeros(n);
    for i in range(len(v)):
        out_v = out_v + c[i] * v[i];
    return out_v
```

```python
c = [1, 2]; 
v = np.array([[1, 0], [0, 1]]);
```

```python
out_vector = lincomb(c, v)
out_vector
```

```python
lincomb([1, 2, 3], v)
```

```python
#Code for the calss Tuesday September 24, 2026;
def f_central(f, x, h = 1e-5):
    return (f(x + h) - f(x - h)) / (2 * h);
```

```python
import numpy as np
f = lambda x: np.exp(x);

ans = f_central(f, 0);
print(ans)
```

```python
d = np.array([1, 2, 3, 4]);
super_d = np.array([-1, -2, -3]);
sub_d = np.array([-100, -200, -300]);

A = np.diag(d) + np.diag(super_d, k = 1) + np.diag(sub_d,  k = -1)
```

```python
print(A)
```

```python
b = np.array([1, 2, 3, 4]);
np.linalg.solve(A, b)
```
