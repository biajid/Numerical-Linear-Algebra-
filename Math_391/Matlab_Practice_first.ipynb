```matlab
x = linspace(0, 2 * pi, 1000);
y1 = sin(x);
y2 = cos(x);
figure('Name', 'Plot of sin function on [0, 2 * pi]');
plot(x, y, x, y2, 'LineWidth', 2);
legend('y1 = sin(x)', 'y2 = cos(x)', 'location', 'best');
hold on;
yline(0, 'LineWidth', 2, 'color', 'k');
hold off;
xlabel('x axis');
ylabel('y axis');
title('Plot of y = sin(x)');
```
