```matlab
% Projectile Motion 
vo = 250; 
g = 9.81;
theta = 65 * pi / 180;
x0 = 3000;
vx = -30;
v_oy = vo * cos(theta);
v_oz = vo * sin(theta);
t = 2 * v_oz / g;
t1 = linspace(0, t, 100);
z = v_oz * t1 - 0.5 * g * t1 .^ 2;
y = v_oy * t1;
x = x0 + vx * t1;
xnowind(1 : length(y)) = x0;
plot3(x, y, z, 'k', xnowind, y, z, 'r');
grid on;
%axis([0 8000 0 8000 0 2500]);
xlabel('x(m)'); ylabel('y(m)'); zlabel('z(m)');
legend('With Wind', 'No Wind')
```

```matlab
% Electric Potential Calculation 
k = 1 ./ (4 * pi * 8.85 * 10 ^ -12);
t = linspace(-0.2, 0.2, 100);
[x, y] = meshgrid(t, t);
q1 = 2e-10;
q2 = 3e-10;
r1 = sqrt((x - 0.25) .^ 2 + y .^ 2);
r2 = sqrt((x + 0.25) .^ 2 + y .^ 2);
v = (1 / k) .* ((q1 ./ r1) + (q2 ./ r2));
mesh(x, y, v);
colorbar;
colormap jet;
xlabel('x');
ylabel('y');
zlabel('Electric Voltage');
```

```matlab
% Heat Conduction Problem 
a = 5;
b = 4;
na = 20;
nb = 16;
k = 5;
T0 = 80;
x = linspace(0, a, na);
y = linspace(0, b, nb);
[x, y] = meshgrid(x, y);

clear T;

for ii = 1 : nb 
    for jj = 1 : na 
        T(ii, jj) = 0;

        for n = 1 : k 
            ns = 2 * n - 1;
            T(ii, jj) = T(ii, jj) + sin(ns * pi * x(ii, jj) / a) .* sinh(ns * pi * y(ii, jj) / a) ...
                ./ ((ns * sinh(ns * pi * b / a)));
        end
        T(ii, jj) = T(ii, jj) * 4 * T0 / pi;
    end
end

surf(x, y, T);
xlabel('x(m)');
ylabel('y(m)');
zlabel('T( ^oC)');


```
