### Power Method Exploration from Gilbert Strang's Linear Algebra and it's Application 4th edition

```matlab
%%file my_power_method.m
function [lambda, v, iter] = my_power_method(A, tol, max_iter, guess_eig)
    % This function will calculate the eigenvalue of a 
    % matrix through shifted inverse power iteration, and would perform better if the matrix is symmetric.
    % usage [lambda, v, iter] = my_power_method(A, tolerance, max_iter, guess_eig)
    % If no guess is given, it will return the smallest eigenvalue.
    if nargin < 4
        guess_eig = 0;
    end
    if nargin < 3
        max_iter = 100;
    end

    if nargin < 2
        tol = 1e-6;
    end
    % I need to add the parent directory to the path if necessary
    addpath("..");
    n = size(A, 1);
    v = rand(n, 1);
    v = v / norm(v);

    
    A_tilda = A - guess_eig * eye(n);
    [L, U] = lu(A_tilda);

    lambda_old = v' * A * v;
    for k = 1 : max_iter
        % I will do shifted power iteration on equation A*v_{k + 1} = v_{k}
        y = L \ v;
        v = U \ y;
        v = v / norm(v);
        lambda = v' * A_tilda * v;

        if abs(lambda - lambda_old) < tol
            break;
        end
        lambda_old = lambda;
    end

    iter = k;

    lambda = v' * A_tilda * v;
    lambda = lambda + guess_eig;
    
end
```

```matlab
A = randn(4, 4) * 100 + rand(4, 4) * 200;
A = (A + A') / 2;
disp(A);
[v, d] = eig(A);
disp("Eigenvalues:");
disp(diag(d));
guess = diag(d);
disp("Eigenvectors:");
disp(v);
```

```matlab
for index = 1 : length(guess)
    [lambda, v, iter] = my_power_method(A, 1e-6, 100, 1.2 * guess(index));
    disp("Eigenvalue:");
    disp(lambda);
    disp("Eigenvector:");
    disp(v);
    disp("Iterations:");
    disp(iter);
end
```
