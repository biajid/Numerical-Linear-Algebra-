```matlab
% Electric dipole modeling
q = 12e-9;
k = 9e9;

x = (-0.05:0.001:0.05)';  % Column vector

rminus = sqrt((x - 0.02).^2 + 0.02^2);
rplus  = sqrt((x + 0.02).^2 + 0.02^2);

eminus = -k * q ./ rminus.^2;
eplus  =  k * q ./ rplus.^2;

% Unit displacement vectors: one [x, y] pair per row
eminusuv = [(x - 0.02) ./ rminus, -0.02 ./ rminus];
eplusuv  = [(x + 0.02) ./ rplus,  -0.02 ./ rplus];

EplusV = [eplusuv(:,1) .* eplus, ...
          eplusuv(:,2) .* eplus];

EminusV = [eminusuv(:,1) .* eminus, ...
           eminusuv(:,2) .* eminus];

Etotal = EplusV + EminusV;
EtotalV = sqrt(Etotal(:,1).^2 + Etotal(:,2).^2);

plot(x, EtotalV, 'k', 'LineWidth', 2);
xlabel('x (m)');
ylabel('|E| (V/m)');
title('Electric Field of a Dipole Along the x-axis');
grid on;
```

```matlab
% Electric field of a dipole modeling along the x-axis
% charge q1 is located at (-0.02, -0.02);
% charge q2 is located at (0.02, -0.02);
k = 9 * 10 ^ 9;
x = -0.05 : 0.001 : 0.05;
qminus = -12e-9;
qplus = 12e-9;
x = x(:);
rminus = sqrt((x - 0.02) .^ 2 + 0.02 ^ 2);;
rplus = sqrt((x + 0.02) .^ 2 + 0.02 ^ 2);;
uvqminus = [((x - 0.02) ./ rminus), (0.02 * ones(length(x), 1) ./ rminus)];
uvqminus(1 : 5, :)
uvqplus = [((x + 0.02) ./ rplus), (0.02 * ones(length(x), 1) ./ rplus)];
eplus = k * qplus ./ rplus .^ 2;
eminus = k * qminus ./ rminus .^ 2;

Evect = eplus .* uvqplus + eminus .* uvqminus;

Evect_magnitude = sqrt(Evect(:, 1) .^ 2 + Evect(:, 2) .^ 2);

plot(x, Evect_magnitude(:), 'r', 'LineWidth', 2);

```
